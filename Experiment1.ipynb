```python
#Sequence Data Type---> String
s = 'Welcome to the Geeks World'
print(s)
print(type(s))
# access string with index
print(s[1])
print(s[-1])
```

```python
#Sequence Data Type---> List
a = [1, 2, 3]
print(a)
b = ["Geeks", "For", "Geeks", 4, 5]
print(b[3])
print(b[-3])
```

```python
#Sequence Data Type---> Tuple
t1 = (1,)
print(type(t1))
t2 = ('Geeks', 'For', 'Geeks', 1, 2)
print(t2[3])
print(t2[-3])
```

```python
#Boolean Data Type
print(type(True))
print(type(False))
```

```python
#Set Data Type
s1 = {"a", "a", "b", "c", "b"}
print(s1)
s2 = {"Geeks", "For", "Geeks"}
for i in s2:
 print(i)
```
